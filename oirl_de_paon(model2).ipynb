```python
# ================================
# Import des bibliothèques
# ================================
import os                      # Gestion des fichiers et dossiers
import cv2                     # Traitement d’images
import numpy as np             # Calcul numérique
import matplotlib.pyplot as plt # Visualisation
from tqdm import tqdm          # Barre de progression

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.model_selection import train_test_split
from sklearn.utils import class_weight

# ================================
# Monter Google Drive
# ================================
from google.colab import drive
drive.mount('/content/drive')

# ================================
# Chemin du dataset
# ================================
folder_path = "/content/drive/MyDrive/projet ia/mini_projet/data"

# Les classes (labels)
classes = ["Healthy", "olive_peacock_spot"]

# ================================
# Chargement des images
# ================================
dataset = []

for class_name in classes:
    class_path = os.path.join(folder_path, class_name)
    label_index = classes.index(class_name)  # 0 ou 1

    # Parcours des images dans chaque dossier
    for img_file in tqdm(os.listdir(class_path)):
        img_path = os.path.join(class_path, img_file)

        img = cv2.imread(img_path)  # Lecture image

        if img is not None:
            img = cv2.resize(img, (224, 224))  # Redimensionnement
            dataset.append((img, label_index)) # Stocker image + label

# ================================
# Séparation données / labels
# ================================
x_images = []
y_images = []

for img, label in dataset:
    x_images.append(img)
    y_images.append(label)

# Conversion en tableau numpy
x = np.array(x_images) / 255.0   # Normalisation (0 → 1)
y = np.array(y_images)

# ================================
# Split train / test
# ================================
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# ================================
# Gestion déséquilibre des classes
# ================================
class_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)

# Transformer en dictionnaire pour Keras
class_weights = dict(enumerate(class_weights))

# ================================
# Data Augmentation
# ================================
datagen = ImageDataGenerator(
    rotation_range=20,     # rotation aléatoire
    zoom_range=0.2,        # zoom
    horizontal_flip=True   # flip horizontal
)

datagen.fit(x_train)

# ================================
# Chargement modèle pré-entraîné
# ================================
base_model = MobileNetV2(
    weights='imagenet',     # poids pré-entraînés
    include_top=False,      # on enlève la partie classification
    input_shape=(224,224,3)
)

# Geler les poids du modèle (feature extraction)
base_model.trainable = False

# ================================
# Construction du modèle
# ================================
x = base_model.output

# Réduction dimensionnelle
x = GlobalAveragePooling2D()(x)

# Couche dense
x = Dense(128, activation='relu')(x)

# Dropout pour éviter overfitting
x = Dropout(0.5)(x)

# Couche de sortie (classification binaire)
predictions = Dense(1, activation='sigmoid')(x)

# Création du modèle final
model = Model(inputs=base_model.input, outputs=predictions)

# ================================
# Compilation
# ================================
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='binary_crossentropy',  # pour classification binaire
    metrics=['accuracy']
)

# Affichage architecture
model.summary()

# ================================
# Early Stopping
# ================================
early_stop = EarlyStopping(
    monitor='val_loss',      # surveiller perte validation
    patience=5,              # attendre 5 epochs sans amélioration
    restore_best_weights=True
)

# ================================
# Entraînement
# ================================
history = model.fit(
    datagen.flow(x_train, y_train, batch_size=16),  # données augmentées
    validation_data=(x_test, y_test),
    epochs=30,
    class_weight=class_weights,  # gestion déséquilibre
    callbacks=[early_stop]
)

# ================================
# Visualisation des résultats
# ================================
plt.figure(figsize=(10,5))

# Accuracy
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='train accuracy')
plt.plot(history.history['val_accuracy'], label='val accuracy')
plt.legend()
plt.title("Accuracy")

# Loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.legend()
plt.title("Loss")

plt.show()
```
